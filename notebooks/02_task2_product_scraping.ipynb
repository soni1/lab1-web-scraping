# Task 2 - Product Scraping

Website: Verkkokauppa.com

The goal is to scrape product names and prices from one product listing page and save the results to a CSV file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.verkkokauppa.com/fi/catalog/phones-and-tablets/puhelimet-tabletit-ja-kellot/products"

response = requests.get(url)

print("Status code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print("Page title:", soup.title.get_text(strip=True))

headings = soup.find_all(["h2", "h3"])
print("Number of h2/h3 headings:", len(headings))

for heading in headings[:10]:
    print(heading.get_text(" ", strip=True))
```

```python
for heading in headings[6:20]:
    print(heading.get_text(" ", strip=True))
```

```python
headings = soup.find_all(["h2", "h3"])

for i, heading in enumerate(headings[:30]):
    text = heading.get_text(" ", strip=True)
    print(i, repr(text))
```

```python
# Inspect links that may contain product names
links = soup.find_all("a")

for i, link in enumerate(links[:80]):
    text = link.get_text(" ", strip=True)

    if text:
        print(i, repr(text))
```

```python
# Look for text containing euro prices
price_texts = soup.find_all(string=lambda text: text and "€" in text)

for i, text in enumerate(price_texts[:30]):
    print(i, repr(text.strip()))
```

```python
url = "https://www.multitronic.fi/puhelimet-ja-tarvikkeet/matkapuhelimet/Android"

response = requests.get(url)

print("Status code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

# Show some visible text that may contain products and prices
for i, tag in enumerate(soup.find_all(["h2", "h3", "a", "span"])[:200]):
    text = tag.get_text(" ", strip=True)

    if text and ("€" in text or len(text) > 20):
        print(i, repr(text))
```

```python
product_links = soup.find_all("a", href=lambda href: href and "/product/" in href)

print("Product links found:", len(product_links))

for i, link in enumerate(product_links[:20]):
    text = link.get_text(" ", strip=True)
    if text:
        print(i, repr(text))
```

```python
product_name = "Realme 16 Pro+"

position = response.text.find(product_name)

print("Position:", position)

if position != -1:
    print(response.text[position - 500:position + 1500])
```

```python
products = soup.find_all("div", class_="productDataEntry")

product_data = []

for product in products:
    name = product.get("data-name")
    price = product.get("data-price")
    currency = product.get("data-currency")

    if name and price:
        product_data.append({
            "Product": name,
            "Price": price,
            "Currency": currency
        })

df = pd.DataFrame(product_data)

print(df.head(10))
print()
print("Products found:", len(df))
```

```python
csv_path = "../output/task2_products.csv"

df.to_csv(csv_path, index=False, encoding="utf-8-sig")

print("Task 2 CSV saved successfully.")
print("Rows saved:", len(df))
```

```python
# Save an Excel-friendly CSV for Finnish/European regional settings.
# Semicolon separates columns and comma is used as the decimal separator.

df_excel = df.copy()

# Convert the Price column from text to numeric values
df_excel["Price"] = pd.to_numeric(df_excel["Price"])

semicolon_csv_path = "../output/task2_products_semicolon.csv"

df_excel.to_csv(
    semicolon_csv_path,
    index=False,
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)

print("Excel-friendly semicolon CSV saved successfully.")
```
