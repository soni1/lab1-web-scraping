# Task 2 - Product Scraping

Website: Multitronic

The goal is to scrape product names and prices from one product listing page and save the results to a CSV file.

## Final selected website: Multitronic

Several e-commerce websites were tested first, but some used dynamic JavaScript loading or rate limiting, which made direct scraping with `requests` unsuitable.

Multitronic was selected because the product information was available directly in the returned HTML and could be extracted reliably.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.multitronic.fi/puhelimet-ja-tarvikkeet/matkapuhelimet/Android"

response = requests.get(url)

print("Status code:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")
```

```python
product_name = "Realme 16 Pro+"

position = response.text.find(product_name)

print("Position:", position)

if position != -1:
    print(response.text[position - 500:position + 1500])
```

```python
products = soup.find_all("div", class_="productDataEntry")

product_data = []

for product in products:
    name = product.get("data-name")
    price = product.get("data-price")
    currency = product.get("data-currency")

    if name and price:
        product_data.append({
            "Product": name,
            "Price": price,
            "Currency": currency
        })

df = pd.DataFrame(product_data)

print(df.head(10))
print()
print("Products found:", len(df))
```

```python
csv_path = "../output/task2_products.csv"

df.to_csv(csv_path, index=False, encoding="utf-8-sig")

print("Task 2 CSV saved successfully.")
print("Rows saved:", len(df))
```

```python
# Save an Excel-friendly CSV for Finnish/European regional settings.
# Semicolon separates columns and comma is used as the decimal separator.

df_excel = df.copy()

# Convert the Price column from text to numeric values
df_excel["Price"] = pd.to_numeric(df_excel["Price"])

semicolon_csv_path = "../output/task2_products_semicolon.csv"

df_excel.to_csv(
    semicolon_csv_path,
    index=False,
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)

print("Excel-friendly semicolon CSV saved successfully.")
```
