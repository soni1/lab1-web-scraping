# Task 1 - Text Scraping

## Topic
Machine Learning

## Websites
1. Stanford HAI
2. Carnegie Mellon University

The goal is to scrape a headline and textual information from two reliable websites and save the collected text to a local text file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Environment ready")
```

```python
url = "https://hai.stanford.edu/ai-definitions/what-is-machine-learning"

response = requests.get(url)

print("Status code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

headline = soup.find("h1")

print(headline.get_text(strip=True))
```

```python
paragraphs = soup.find_all("p")

stanford_paragraphs = []

for p in paragraphs:
    text = p.get_text(" ", strip=True)

    if len(text) > 150 and text not in stanford_paragraphs:
        stanford_paragraphs.append(text)

stanford_text = stanford_paragraphs[0]

print("Headline:", headline.get_text(strip=True))
print()
print(stanford_text)
```

```python
cmu_url = "https://www.ml.cmu.edu/"

cmu_response = requests.get(cmu_url)

print("Status code:", cmu_response.status_code)
```

```python
cmu_soup = BeautifulSoup(cmu_response.text, "html.parser")

cmu_headline = cmu_soup.find("h1")
cmu_paragraphs = cmu_soup.find_all("p")

print("Headline:", cmu_headline.get_text(strip=True))

cmu_clean_paragraphs = []

for p in cmu_paragraphs:
    text = p.get_text(" ", strip=True)

    if len(text) > 120 and text not in cmu_clean_paragraphs:
        cmu_clean_paragraphs.append(text)

cmu_text = cmu_clean_paragraphs[0]

print()
print(cmu_text)
```

```python
output_text = f"""TASK 1 - TEXT SCRAPING
Topic: Machine Learning

SOURCE 1: Stanford HAI
Headline: {headline.get_text(strip=True)}
Text:
{stanford_text}

SOURCE 2: Carnegie Mellon University
Headline: {cmu_headline.get_text(strip=True)}
Text:
{cmu_text}
"""

with open("../output/task1_text.txt", "w", encoding="utf-8") as file:
    file.write(output_text)

print("Task 1 saved successfully.")
```
